# BPM — Beat Per Minute: Theory, Common Tempos & Interactive Practice Tool

**Beats per minute (BPM)** is the universal unit for musical tempo — the speed at which a piece of music moves. One "beat" is typically one quarter-note in standard 4/4 time, and the count of how many of those beats fit in 60 seconds is the BPM.

$$\text{BPM} = \frac{60}{\text{beat duration (seconds)}}$$

This notebook covers:
1. The physics of a beat pulse
2. Classical Italian tempo markings and their BPM ranges
3. Genre-by-genre BPM landscape across modern music
4. Visualizations: waveform, BPM ruler, inter-beat interval (IBI)
5. An embedded **interactive web metronome** — tap tempo, visual flash, Web Audio click

## 1  What Is a Beat?

A beat is the perceptual "tick" that listeners synchronise to when they tap their foot. Physically, a metronome click is a short broadband pulse — essentially an impulse function convolved with the resonance of the instrument (or a synthesised sine burst).

Key relationships:

| Quantity | Formula | Example at 120 BPM |
|---|---|---|
| Beat period (s) | \(T = 60 / \text{BPM}\) | 0.500 s |
| Beat frequency (Hz) | \(f = \text{BPM} / 60\) | 2.000 Hz |
| 8th-note period (s) | \(T/2\) | 0.250 s |
| 16th-note period (s) | \(T/4\) | 0.125 s |
| Bar (4/4) duration (s) | \(4T\) | 2.000 s |

> **Inter-beat interval (IBI)** is the inverse of BPM frequency. Heart-rate monitors and music analysis software both use IBI. Normal resting heart rate (60–100 BPM) maps to IBI of 0.6–1.0 s — the same range as walking tempo!

## 3  Classical Italian Tempo Markings

Before metronomes existed (Maelzel patented his in 1815), composers used Italian words to describe tempo. Modern editions typically attach BPM ranges, though these are **conventions**, not laws.

| Marking | Meaning | Typical BPM |
|---|---|---|
| Larghissimo | Extremely slow | < 24 |
| Largo | Broad, stately | 24 – 54 |
| Larghetto | Somewhat broad | 54 – 60 |
| Adagio | Slow, at ease | 54 – 66 |
| Adagietto | Somewhat slow | 66 – 76 |
| Andante | Walking pace | 76 – 108 |
| Andantino | Slightly faster than Andante | 80 – 108 |
| Moderato | Moderate | 86 – 120 |
| Allegretto | Moderately fast | 96 – 120 |
| Allegro | Fast, lively | 120 – 168 |
| Vivace | Lively and quick | 132 – 168 |
| Presto | Very fast | 168 – 200 |
| Prestissimo | Extremely fast | > 200 |

```python
tempo_markings = [
    ('Larghissimo',  0,   24),
    ('Largo',       24,   54),
    ('Larghetto',   54,   60),
    ('Adagio',      54,   66),
    ('Adagietto',   66,   76),
    ('Andante',     76,  108),
    ('Andantino',   80,  108),
    ('Moderato',    86,  120),
    ('Allegretto',  96,  120),
    ('Allegro',    120,  168),
    ('Vivace',     132,  168),
    ('Presto',     168,  200),
    ('Prestissimo',200,  280),
]

cmap = plt.cm.coolwarm
fig, ax = plt.subplots(figsize=(11, 5))

for i, (name, lo, hi) in enumerate(tempo_markings):
    color = cmap((lo + hi) / 2 / 280)
    ax.barh(i, hi - lo, left=lo, height=0.7, color=color, alpha=0.85, edgecolor='white')
    ax.text((lo + hi) / 2, i, name, ha='center', va='center', fontsize=8.5,
            fontweight='bold', color='white')

ax.set_yticks(range(len(tempo_markings)))
ax.set_yticklabels([f'{lo}–{hi}' for _, lo, hi in tempo_markings], fontsize=8)
ax.set_xlabel('BPM')
ax.set_title('Classical Italian Tempo Markings — BPM Ranges')
ax.set_xlim(0, 290)
ax.xaxis.set_minor_locator(MultipleLocator(10))
ax.grid(axis='x', which='major', alpha=0.3)
ax.grid(axis='x', which='minor', alpha=0.1)
plt.tight_layout()
plt.show()
```

| Genre | Typical BPM | Notes |
|---|---|---|
| Ballad / Slow Soul | 60 – 80 | Often with rubato |
| Hip-Hop / Trap | 60 – 100 | Often half-time feel; trap sits 130–150 halved |
| R&B / Neo-Soul | 75 – 110 | |
| Rock (mid-tempo) | 90 – 130 | |
| Pop | 100 – 130 | Sweet spot ~116 BPM |
| Reggaeton | 90 – 100 | Dembow rhythm |
| House | 120 – 130 | 4-on-the-floor |
| Techno | 130 – 160 | |
| Trance | 128 – 145 | |
| Drum & Bass | 160 – 180 | Often 174 BPM |
| Hardcore / Gabber | 160 – 300 | |
| Bluegrass | 90 – 200 | Wide, instrument-dependent |

## 5  Visualising a Beat Waveform

A typical synthesised metronome click is a **short sine burst** — here at 1000 Hz, 10 ms duration, tapered with a Hann window to avoid clicks from sharp edges. We'll render four beats at 120 BPM.

```python
SR = 44100          # sample rate Hz
BPM = 120
T = 60 / BPM       # beat period in seconds
N_BEATS = 4
CLICK_FREQ = 1000  # Hz
CLICK_DUR = 0.012  # seconds

total_samples = int(N_BEATS * T * SR)
audio = np.zeros(total_samples)

click_len = int(CLICK_DUR * SR)
t_click = np.linspace(0, CLICK_DUR, click_len, endpoint=False)
click = np.sin(2 * np.pi * CLICK_FREQ * t_click) * np.hanning(click_len)

for beat in range(N_BEATS):
    start = int(beat * T * SR)
    audio[start:start + click_len] += click

t = np.arange(total_samples) / SR

fig, axes = plt.subplots(2, 1, figsize=(11, 5), sharex=True,
                         gridspec_kw={'height_ratios': [3, 1]})

axes[0].plot(t, audio, color='steelblue', linewidth=0.6)
axes[0].set_ylabel('Amplitude')
axes[0].set_title(f'4 Metronome Clicks at {BPM} BPM  (SR = {SR} Hz, click = {CLICK_FREQ} Hz sine burst)')
for b in range(N_BEATS):
    axes[0].axvline(b * T, color='tomato', lw=1, ls='--', alpha=0.7,
                    label='beat' if b == 0 else '')
axes[0].legend(fontsize=9)

# Bottom strip: beat marker
axes[1].set_facecolor('#f5f5f5')
for b in range(N_BEATS):
    axes[1].barh(0, 0.04, left=b * T, height=1, color='tomato', alpha=0.8)
axes[1].set_yticks([])
axes[1].set_xlabel('Time (s)')
axes[1].set_ylabel('Beat\nMarker', fontsize=8)
axes[1].set_xlim(0, N_BEATS * T)

plt.tight_layout()
plt.show()
```

## 6  The BPM Ruler — Note Durations at Any Tempo

A practical "ruler" showing how long each note subdivision lasts for a set of common tempos.

```python
bpms = [60, 80, 100, 120, 140, 160, 180]
divisions = {
    'Whole note (1)': 4,
    'Half note (1/2)': 2,
    'Quarter note (1/4)': 1,
    '8th note (1/8)': 0.5,
    '16th note (1/16)': 0.25,
    '32nd note (1/32)': 0.125,
}

print(f"{'Note':<22}" + "".join(f"{b:>9} BPM" for b in bpms))
print("-" * (22 + 12 * len(bpms)))
for note, beats in divisions.items():
    row = f"{note:<22}"
    for b in bpms:
        dur = beats * 60 / b
        row += f"  {dur:>7.4f} s"
    print(row)
```

```python
fig, ax = plt.subplots(figsize=(11, 5))

colors = plt.cm.tab10(np.linspace(0, 1, len(divisions)))
x = np.arange(len(bpms))
width = 0.13
offsets = np.linspace(-(len(divisions)-1)/2, (len(divisions)-1)/2, len(divisions)) * width

for (note, beats), color, offset in zip(divisions.items(), colors, offsets):
    durs = [beats * 60 / b for b in bpms]
    bars = ax.bar(x + offset, durs, width, label=note, color=color, alpha=0.85)

ax.set_xticks(x)
ax.set_xticklabels([str(b) for b in bpms])
ax.set_xlabel('BPM')
ax.set_ylabel('Duration (s)')
ax.set_title('Note Duration (s) for Each BPM & Subdivision')
ax.legend(fontsize=8, ncol=2)
ax.set_yscale('log')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.3f}'))
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 7  Half-Time & Double-Time Relationships

Many genres exploit the ambiguity between a groove and its double. A track at **170 BPM** with a half-time snare on beat 3 feels like **85 BPM** to a casual listener. Conversely, a 90 BPM Hip-Hop beat can be reframed as 180 BPM with sixteenth-note hi-hats.

```python
fig, ax = plt.subplots(figsize=(10, 4))

bpm_vals = np.arange(40, 300, 1)
half = bpm_vals / 2
double = bpm_vals * 2

ax.plot(bpm_vals, bpm_vals, color='gray', lw=1, ls='--', label='1× (identity)')
ax.plot(bpm_vals, half, color='steelblue', lw=2, label='Half-time (÷2)')
ax.plot(bpm_vals, np.clip(double, 0, 400), color='tomato', lw=2, label='Double-time (×2)')

# Shade region where double-time > 300 (unusable territory)
mask = double > 300
ax.fill_between(bpm_vals, 300, np.clip(double, 0, 400), where=mask,
                color='tomato', alpha=0.08, label='>300 BPM (extreme)')

ax.set_xlim(40, 300)
ax.set_ylim(0, 410)
ax.set_xlabel('Notated BPM')
ax.set_ylabel('Perceived / alternate BPM')
ax.set_title('Half-Time & Double-Time Feel')
ax.legend()
ax.grid(alpha=0.3)

# Example annotation: Drum & Bass 174 / half-time 87
ax.annotate('D&B 174 BPM\n↔ half-time 87',
            xy=(174, 87), xytext=(190, 120),
            arrowprops=dict(arrowstyle='->', color='steelblue'),
            color='steelblue', fontsize=9)
plt.tight_layout()
plt.show()
```

## 9  Tap Tempo Math

The "tap tempo" algorithm is simple: record the timestamp of each tap and compute the average inter-tap interval over the last *N* taps.

$$\text{BPM} = \frac{60{,}000}{\overline{\Delta t_{\text{ms}}}}$$

Using more taps (larger *N*) reduces jitter but increases latency before converging on the correct tempo. 4–6 taps is typical in commercial apps.

```python
rng = np.random.default_rng(42)
TRUE_BPM = 120
TRUE_IBI = 60000 / TRUE_BPM  # ms
N_TAPS = 16
JITTER_SD = 30  # ms human tap jitter (realistic)

tap_times = np.cumsum(TRUE_IBI + rng.normal(0, JITTER_SD, N_TAPS))
estimated = []
for n in range(2, N_TAPS+1):
    diffs = np.diff(tap_times[:n])
    estimated.append(60000 / diffs.mean())

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(range(2, N_TAPS+1), estimated, marker='o', color='steelblue', markersize=5)
ax.axhline(TRUE_BPM, color='tomato', ls='--', lw=1.5, label=f'True BPM = {TRUE_BPM}')
ax.fill_between(range(2, N_TAPS+1), TRUE_BPM-2, TRUE_BPM+2,
                color='tomato', alpha=0.1, label='±2 BPM tolerance')
ax.set_xlabel('Number of taps used')
ax.set_ylabel('Estimated BPM')
ax.set_title(f'Tap Tempo Convergence (True={TRUE_BPM} BPM, jitter σ={JITTER_SD} ms)')
ax.legend()
ax.grid(alpha=0.3)
ax.set_xlim(2, N_TAPS)
plt.tight_layout()
plt.show()
```

## 10  Key Takeaways

| Concept | Summary |
|---|---|
| **BPM formula** | \(\text{BPM} = 60 / T\), where \(T\) is the beat period in seconds |
| **Classical markings** | Loosely standardised; Andante ≈ 76–108, Allegro ≈ 120–168 |
| **Genre clusters** | Hip-Hop 60–100, Pop 100–130, House ~128, D&B 160–180 |
| **Half-time feel** | A 170 BPM D&B track *feels* like 85 BPM with a half-time groove |
| **Note durations** | Halving BPM doubles every note's physical duration |
| **Tap tempo** | Averages inter-tap intervals; converges within 4–6 taps |
| **Web Audio API** | Enables sub-millisecond accurate scheduling of click events in-browser |
| **Metronome click** | Short sine burst (~10 ms), Hann-windowed to avoid spectral leakage / distortion |
